```python
"""
python Regular Expression
1.Recognize specific words → bat, bit, but, hat, hit, hut
2.Match two words with one space → John Smith
3.Match surname + initial → Smith, J
4.Match a Python identifier → student_name, _value, name123
5.Match a street address → 1180 Bordeaux Drive
6.Match a simple website domain → www.google.com
7.Match a URL using a loose pattern → https://google.com
8.Match a URL using a stricter pattern
9.Extract a type name → get int from <type 'int'>
10.Match a phone number → 800-555-1212 or 555-1212
11.Match a phone number with different area-code formats → (800) 555-1212
"""
"""
Specific words      → [bh][aiu]t
Two words           → word space word
Last, Initial       → word, letter
Python identifier   → letter/_ + letters/digits/_
Street address      → number + words
WWW domain          → www.name.com
Loose URL           → http/https://anything
Tight URL           → protocol + domain + extension
Extract type        → capturing group
Phone               → optional area + 3-4 digits
Phone parentheses   → optional (area) + 3-4 digits
"""
```

```python
import re

# 1. bat, bit, but, hat, hit, hut
pattern = r"^[bh][aiu]t$"

text = "bat"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 2. Pair of words separated by a single space
pattern = r"^[A-Za-z]+ [A-Za-z]+$"

text = "John Smith"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 3. Last name, first initial
pattern = r"^[A-Za-z]+, [A-Za-z]$"

text = "Smith, J"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 4. Valid Python identifier
pattern = r"^[A-Za-z_][A-Za-z0-9_]*$"

text = "student_name"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 5. Street address
pattern = r"^\d+ [A-Za-z]+(?: [A-Za-z]+)*$"

text = "1180 Bordeaux Drive"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 6. Simple web domain
pattern = r"^www\.[A-Za-z0-9-]+\.com$"

text = "www.google.com"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 7. URL - loose pattern
pattern = r"^https?://.+$"

text = "https://google.com"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 8. URL - tightened pattern
pattern = r"^https?://(?:www\.)?[A-Za-z0-9-]+\.[A-Za-z]{2,}(?:/[^\s]*)?$"

text = "https://www.google.com/page"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 9. Extract type name from <type 'int'>
pattern = r"<type '([^']+)'>"

text = "<type 'int'>"

match = re.search(pattern, text)

if match:
    print("Type:", match.group(1))
else:
    print("Not Matched")
```

```python
# 10. Phone number with optional area code
pattern = r"^(?:\d{3}-)?\d{3}-\d{4}$"

text = "800-555-1212"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```

```python
# 11. Phone number with optional hyphenated or parenthesized area code
pattern = r"^(?:\d{3}-|\(\d{3}\) )?\d{3}-\d{4}$"

text = "(800) 555-1212"

if re.fullmatch(pattern, text):
    print("Matched")
else:
    print("Not Matched")
```
